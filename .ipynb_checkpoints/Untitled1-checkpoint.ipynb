```python
import requests
import json

url = "http://m1.tme-crypto.fr:8888/"
headers = {
    'Content-Type': 'application/json',
}
```

Item affiché sur le tableau a l'atrium:

Normalement affiche: `{"jsonrpc": "2.0", "id": 737, "result": "TELEPORT_MODULE"}`

```python
data = {
    "jsonrpc": "2.0",
    "method": "item.name",
    "params": {"world_id": "807b4a5d685915218c7987708c110cbe", "item":"4c2d59f902b1a1dd8f62f9526153d447"},
    "id": 737
}
response = requests.post(url, headers=headers, data=json.dumps(data))
print(response.text)
```
