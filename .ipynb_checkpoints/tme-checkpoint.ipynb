````python
def decrypt(cipherText, passphrase):
    if isinstance(cipherText, str):
        cipherText = cipherText.encode('utf-8')

    cipherText = base64.b64decode(cipherText)

    with open("encrypted_message.bin", "wb") as f:
        f.write(cipherText)

    subprocess.run([
        "openssl", "enc", "-d", "-aes-128-cbc", "-pbkdf2", 
        "-in", "encrypted_message.bin", 
        "-out", "decrypted_message.txt", 
        "-pass", f"pass:{passphrase}"
    ], check=True)

    with open("decrypted_message.txt", "rb") as f:
        decrypted_message = f.read()
    
    # Clean up the files
    os.remove("encrypted_message.bin")
    os.remove("decrypted_message.txt")
    
    return decrypted_message
````

````python
def encrypt_RSA(string_to_encrypt):
    """
    Le message clair DOIT être plus court que la clef.  Si ce n'est pas le cas,
   l'opération échoue avec un message "Public Key operation error, data too 
   large for key size".  Avec une clef de 2000 bits, vous avez droit à environ
   1800 bits de message.  C'est une limitation de l'algorithme lui-même.
   """
    
    if isinstance(string_to_encrypt, str):
        string_to_encrypt = string_to_encrypt.encode('utf-8')
        
    args = [
        'openssl',
        'pkeyutl',
        '-encrypt',
        '-pubin',
        '-inkey',
        "pki.pem"
    ]
    
    result = subprocess.run(args, input=string_to_encrypt, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    
    error_message = result.stderr.decode()
    if error_message != '':
        raise OpensslError(error_message)
        
    return result.stdout
````

Challenge:          wowee sinew azure zincs mahua

````python
def signer_challenge(challenge, private_key_path):
    with open("challenge.txt", "w") as ch:
        ch.write(challenge)

    sign = "signature"
    
    args = [
        'openssl',
        'dgst',
        '-sha256',
        '-sign',
        private_key_path,
        '-out',
        sign,
        "challenge.txt"
    ]
    
    subprocess.run(args, check=True)

    with open((sign), 'rb') as out:
        signature = out.read()

    os.remove(sign)
    os.remove("challenge.txt")
    
    return signature.hex()
````

````python
from hashlib import sha256
def key_expansion(seed : bytes) -> bytes:
    """
    Renvoie 256 bits pseudo-aléatoires à partir de seed
    """
    state = seed
    output = b''
    for i in range(8):
        state = sha256(state).digest()
        output += state[:4]
    return output
````

````python
def get_seed(iv_cible):
    for i in range(2**16):
        seed = i.to_bytes(2, 'big')
        iv = bytes.fromhex(iv_cible)
        if key_expansion(seed)[16:] == iv:
            return seed
    return None

seed = get_seed(iv_cible)
print("Seed:", seed.hex())

key = key_expansion(seed)
K = key[:16]
IV = key[16:]
print("Key:", K.hex())
print("IV:", IV.hex())
````

````python
def decrypt_with_key_iv(plaintext, key, iv):
    if isinstance(plaintext, str):
        plaintext = plaintext.encode('utf-8')
    
    key_hex = key.hex()
    iv_hex = iv.hex()
    
    result = subprocess.run([
        'openssl', 'enc','-d', '-aes-128-cbc', '-K', key_hex, '-iv', iv_hex, '-base64'
    ], input=plaintext, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    
    error_message = result.stderr.decode()
    if error_message:
        raise OpensslError(error_message)
    
    return result.stdout.decode().strip()


encrypted_message = decrypt_with_key_iv(message, K, IV)
print("Encrypted message:\n", encrypted_message)
````

````python
def creer_certificat(file_key, username):
    args = [
        'openssl',
        'req',
        '-new',
        '-key',
        file_key,
        '-batch',
        '-subj',
        f"/CN={username}"
    ]
    
    result = subprocess.run(args, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    
    error_message = result.stderr.decode()
    if error_message != '':
        raise OpensslError(error_message)
        
    return result.stdout
````

````python
import cryptography.hazmat.primitives.serialization
from hashlib import sha256
import random
import sympy

# Function names respect those in https://www.ietf.org/rfc/rfc3447.txt
# SHA-256
HASH_ID = b'010\r\x06\t`\x86H\x01e\x03\x04\x02\x01\x05\x00\x04 '

def i2osp(x : int, k : int) -> bytes:
    """
    Convert the integer x to a sequence of k bytes
    """
    return x.to_bytes(k, byteorder='big')

def os2ip(x : bytes) -> int:
    """
    Convert the sequence of bytes to an integer
    """
    return int.from_bytes(x, byteorder='big')

def emsa_pkcs1_encode(M : bytes, k : int) -> bytes:
    """
    Encode a message into k bytes for RSA signature
    """
    h = sha256(M)
    T = HASH_ID + h.digest()
    if len(T) + 11 > k:
        raise ValueError("Message Too Long")
    PS = bytes([0xff] * (k - len(T) - 3))
    EM = bytes([0x00, 0x01]) + PS + bytes([0x00]) + T
    return EM

def emsa_pkcs1_decode(EM : bytes, k : int) -> bytes:
    """
    Given an EMSA_PKCS1-encoded message, returns the Hash

    >>> x = emsa_pkcs1_encode("toto", 128)
    >>> emsa_pkcs1_decode(x, 128) == sha256("toto".encode()).digest()
    True
    """
    if len(EM) != k:
        raise ValueError("Incorrect Size")
    if EM[:2] != bytes([0x00, 0x01]):
        raise ValueError("Incorrect Header")
    i = 2
    while EM[i] != 0:
        if EM[i] != 0xff:
            raise ValueError("Incorrect Filler")
        i += 1
        if i == k:
            raise ValueError("Only Filler")
    if i < 10:
        raise ValueError("Not enough filler")
    T = EM[i+1:]
    if T[:len(HASH_ID)] != HASH_ID:
        raise ValueError("Bad Hash ID")
    H = T[len(HASH_ID):]
    return H

def key_length(n : int) -> int:
    """
    key length in bytes
    """
    return (n.bit_length() + 7) // 8

def rsa_pkcs_sign(n : int, d : int, M : bytes):
    """
    RSA Signature using PKCS#1 v1.5 encoding
    """
    k = key_length(n)
    EM = emsa_pkcs1_encode(M, k)
    m = os2ip(EM)
    s = pow(m, d, n)
    S = i2osp(s, k)
    return S

def rsa_pkcs_verify(n : int, e : int, M : bytes, S : bytes) -> bool:
    """
    Verify RSA PKCS#1 v1.5 signatures
    """
    k = key_length(n)
    if len(S) != k:
        raise ValueError("Bad length")
    s = os2ip(S)
    m = pow(s, e, n)
    EM = i2osp(m, k)
    H = emsa_pkcs1_decode(EM, k)
    return (H == sha256(M).digest())
````

## Flag 13

Aller dans la salle de TME, pour effectuer le TME 1

```
>>> conseil poste 1
Pour que g puisse être d'ordre q, il faut que p-1 soit un multiple de q.
Il faut donc fabriquer p en vérifiant que
        1 + q*(nombre aléatoire pair de la bonne taille)
est bien un nombre premier.


>>> conseil poste 1
Pour trouver g, ce qu'il ne faut PAS faire, c'est essayer des valeurs aléatoires.
Ça n'a aucune chance de marcher.  Mais il y a un moyen simple de produire des 
nombres qui sont d'ordre q, s'ils sont différents de 1.


>>> conseil poste 1
Le truc consiste à choisir un nombre x aléatoirement, puis à calculer
        g := x ** ((p-1) // q)     [modulo p]
Si g == 1, recommencer (c'est très improbable).
Sinon, on a :
        g ** q == x ** (p-1)       [modulo p]
               == 1                (d'après le petit théorème de Fermat)

```

````python
import random
from math import gcd
import sympy

def trouver_p_premier(a, b, q):
    while True:
        # Générer un nombre aléatoire k de taille appropriée
        k = random.randint((a-1)//q, (b-1)//q)
        
        # Assurer que k est pair pour que p-1 soit divisible par q
        if k % 2 == 1:
            k += 1
        
        # Calculer p = 1 + q*k
        p = 1 + q*k
        
        # Vérifier si p est dans l'intervalle [a, b[
        if p < a or p >= b:
            continue
            
        # Vérifier si p est premier
        if sympy.isprime(p):
            return p
def trouver_generateur_g(p, q):
    while True:
        # Choisir un nombre aléatoire x entre 2 et p-2
        x = random.randint(2, p-2)
        
        # Calculer g = x^((p-1)/q) mod p
        g = pow(x, (p-1) // q, p)
        
        # Si g != 1, alors g est d'ordre q
        if g != 1:
            return g
def generer_p_g(a, b, q):
    p = trouver_p_premier(a, b, q)
    g = trouver_generateur_g(p, q)
    return p, g
````

## Flag 14

```
>>> conseil tme #2
Il est facile de trouver un g qui marche... si on dispose de la factorisation
de p-1.  Le problème principal consiste donc à produire un nombre premier p 
tout en connaissant la factorisation de p-1.  La seule procédure envisageable 
consiste à produire des nombres X de la bonne taille dont la factorisation est 
connue, puis à tester si X+1 est premier.


>>> conseil tme #2
En plus, on nous impose un facteur premier q qui doit diviser p-1.
La technique la plus simple, mais probablement pas la plus rapide, pour 
fabriquer p, consiste à générer un grand nombre premier r, puis à poser
p = 2qr + 1.  Si p est premier, alors la factorisation de p-1 est connue :
les diviseurs premiers de p-1 sont 2, q et r.


>>> conseil tme #2
Pour faire marcher cette technique, il faut générer des nombres aléatoires r 
dans le bon intervalle, puis tester s'ils sont premier, avant de tester si
le p auquel cela aboutit est premier.  Autrement dit, pour avoir
        a <= 1 + 2qr < b,
il faut poser
        (a - 1) / 2q <= r < (b - 1) / 2q.


>>> conseil tme #2
Il est possible de faire plus rapide. La rumeur prétend que certains h4x0rz ont
écrit du code Python pur qui résoud le problème en 5 secondes environ.
Pour aller plus vite, il faut éviter au maximum de faire des tests de primalité
sur des grands entiers.  Pour cela, il faut fabriquer p-1 avec plusieurs petits
facteurs premiers, puis tester si p est premier.


>>> conseil tme #2
Pour produire k nombres premiers p_1, ..., p_k de telle sorte que 
        Q = p_1 * p_2 * ... * p_k 
appartienne à l'intervalle [a; b[, il n'y a pas 36 solutions.  En gros, la
technique est la suivante : 
1) Fabriquer dans un premier temps les k-1 premiers nombres librement
2) Fabriquer le dernier en faisant bien attention à ce que produit soit
   dans [a; b[
Après la phase 1, on se trouve avec Q' = p_1 * p_2 * ... * p_{k-1}. Pour que le
produit final soit dans le bon intervalle, il faut choisir p_k dans l'intervalle
[a/Q'; b/Q'[.


>>> conseil tme #2
Si on met en oeuvre la stratégie précédente, il y a un problème à éviter :   
plus Q' est grand, plus l'intervalle [a/Q'; b/Q'[ rétrecit.  Et s'il est
trop petit, il peut ne pas contenir de nombre premier du tout (impossible
de trouver p_k dedans, donc).  Ou bien, il peut ne pas en contenir suffisamment
(possible de trouver p_k, mais p = Q + 1 n'est jamais premier).  Il est 
difficile d'obtenir des certitudes, mais il est raisonnable de penser qu'il 
faut qu'il contienne O(|b|) nombres différents pour que l'un d'entre eux soit 
premier avec bonne probabilité.  Il faut donc que 
        (b/Q') - (a/Q') >= [constante arbitraire] * |b|**2.
Pour s'en sortir, il faut donc :
- déterminer dans un premier temps la taille que doit avoir Q' en bits.
- produire Q' en multipliant ensemble des nombres premiers arbitraires.
- trouver le dernier facteur premier.


```

````python
a = int("1fe87379090e0b7a5ea101a856f900edeb74d12c38196acf7ecbe3b428b5c74bed5b77a3db15e4eeb204769353bc6091ce80eec30ef393adf4d7810c9c66a0b2b5eb6f1bf3ffbb4e01d222960825c0d280558f56044d993589e03f3538bed63913199f5042714a798b9e424350d01b942ecf2fd6af68a16b385d85de8200bf35717503d7d17d4db55b43fda97b64fd9dcaae2f2d5fcdbc0a59045f5613b6c2dc8eff55792d7258738895a23e949c1740f0c7c3debf713c84c19c1e6d7d151ac426435a0361427301641de2e302ae2ffbea585b4b29ddbe4b07a8ab44eee78368cfae46c7e189ba42813d4ce0a015fa79e7e31794a085d4879263ceae6229cbfb",16)
b = a + 2**1950
q = int("6313505359c12637e70cc3876863168196c36f17cbd65b834c8581c056d637672e29698c47edb3e8fe42279f20d75707",16)

from random import randint

def genere_q_prime(a,b,q) :
    # Q'
    Q_prime = (b-a)//(len(bin(b))**2)
    # La taille en bit :
    Q_bit = len(bin(Q_prime))

    q_prime = q 
    p_k = 2

    list_pk = []
    # On multiplie les premiers jusqu'à avoir un nombre de la taille en bit de q'
    while len(bin(q_prime)) < Q_bit :
        # On met à jour q'
        q_prime = q_prime*p_k
        # On l'ajoute à la liste des facteurs
        list_pk.append(p_k)
        # On va chercher le premier suivant
        p_k = nextprime(p_k)

    # On rend la liste des premiers et q'
    return list_pk,q_prime

# Fonction pour générer le premier p en fonction de q
def genere_p(a,b,q,list_pk) :
    """
        q = q' = produit de p_1, . . ., p_k
    """
    p = -1 
    
    # On s'arrete seulement quand on aura trouver le dernier facteur K
    while True :

        # On prend un nombre aléatoire entre a/q' et b/q'
        K = randint((a-1)//(q),(b-1)//(q))
        # On calcul la valeur de p
        p = 1+q*K

        # est-ce que K est premier ?
        if isprime(K) :

            # Est ce que p est premier
            if isprime(p) :

                # On ajoute K à la liste des facteurs
                list_pk.append(K)
                # On affiche
                print("K =",K)
                # Fin de la boucle 
                break
    # On renvoie p
    return p, K

# Fonction pour générer g d'ordre q mod p
def genere_g(p, fac_list):
    for g in range(2, p):
        if all(pow(g, (p - 1) // f, p) != 1 for f in fac_list):
            return g
````

````python
import base64
import textwrap
import secrets
from hashlib import sha256
from binascii import unhexlify

class LamportKey:
    """
    Abstract Base Class for Lamport Keys (both public and private)
    """
    def __init__(self, key : bytes):
        self.key = [key[i:i+32] for i in range(0, 16384, 32)]

    @classmethod
    def _prefix(cls):
        return f"-----BEGIN {cls.KIND}-----"

    @classmethod
    def _suffix(cls):
        return f"-----END {cls.KIND}-----"

    def dumps(self) -> str:
        """
        Return a PEM representation of the key
        """
        payload = base64.b64encode(b''.join(self.key)).decode()
        middle = '\n'.join(textwrap.wrap(payload, width=64))
        return f"{self._prefix()}\n{middle}\n{self._suffix()}"

    @classmethod
    def loads(cls, key : str):
        """
        Return a key object from its string representation
        """
        prefix = cls._prefix()
        suffix = cls._suffix()
        if not key.startswith(prefix):
            raise ValueError("not a PEM-encoded key (missing prefix)")
        if not key.endswith(suffix):
            raise ValueError("not a PEM-encoded key (missing suffix)")
        payload = key[len(prefix):-len(suffix)]
        return cls(base64.b64decode(payload))

class LamportPrivateKey(LamportKey):
    KIND = 'LAMPORT SECRET KEY'

    @staticmethod
    def keygen():
        """
        Generate a fresh random private key
        """
        return LamportPrivateKey(secrets.token_bytes(512*32))

    def pk(self):
        """
        Return the public key associated to this private key
        """
        pk = b''
        for i in range(512):
            pk += sha256(self.key[i]).digest()
        return LamportPublicKey(pk)

    def sign(self, m : bytes) -> str:
        """
        Sign a message using this private key
        """
        sig = b''
        h = int.from_bytes(sha256(m).digest(), byteorder='big')
        for i in range(256):
            b = (h >> i) & 1
            sig += self.key[2 * i + b]
        return sig.hex()

class LamportPublicKey(LamportKey):
    KIND = 'LAMPORT PUBLIC KEY'

    def verify(self, m : bytes, sig : str) -> bool:
        """
        Verify a signature using this public key
        """
        bsig = bytes.fromhex(sig)
        s = [bsig[i:i+32] for i in range(0, 8192, 32)]
        h = int.from_bytes(sha256(m).digest(), byteorder='big')
        for i in range(256):
            b = (h >> i) & 1
            if sha256(s[i]).digest() != self.key[2 * i + b]:
                return False
        return True
key = LamportPrivateKey.keygen()
msg = b'hello world'
sig = key.sign(msg)
public_key = key.pk()
public_key.verify(msg, sig)
````

````python
import json
from hashlib import sha256

def recover_private_key(signatures_data):
    """
    Récupère la clé privée à partir de plusieurs signatures Lamport
    
    Args:
        signatures_data: Liste de tuples (message, signature)
    
    Returns:
        La clé privée reconstruite (ou partiellement reconstruite)
    """
    # Initialisation de la clé privée récupérée avec 512 valeurs None
    recovered_key = [None] * 512
    
    for message_bytes, signRécupéré 512/512 parties de la clé privée (100.00%)
Key part 0: 8ature_hex in signatures_data:
        # Convertir la signature hex en bytes
        signature_bytes = bytes.fromhex(signature_hex)
        # Diviser la signature en blocs de 32 octets
        signature_blocks = [signature_bytes[i:i+32] for i in range(0, len(signature_bytes), 32)]
        
        # Calculer le hash du message
        hash_value = int.from_bytes(sha256(message_bytes).digest(), byteorder='big')
        
        # Pour chaque bit du hash
        for i in range(256):
            bit = (hash_value >> i) & 1
            # Récupérer la partie correspondante de la clé privée
            key_index = 2 * i + bit
            recovered_key[key_index] = signature_blocks[i]
    
    # Compter combien de parties de la clé ont été récupérées
    recovered_count = sum(1 for k in recovered_key if k is not None)
    print(f"Récupéré {recovered_count}/512 parties de la clé privée ({recovered_count/512*100:.2f}%)")
    
    return recovered_key

def load_signatures_from_file(filename):
    """Charge les signatures depuis un fichier JSON"""
    with open(filename, 'r') as f:
        data = json.load(f)
    
    signatures = []
    for entry in data:
        message = bytes(entry['message'], "utf-8") if isinstance(entry['message'], str) else entry['message'].encode()
        signatures.append((message, entry['signature']))
    
    return signatures

# Exemple d'utilisation
def main():
    filename = "catalogue.json"  # Changez ceci pour le nom de votre fichier
    signatures = load_signatures_from_file(filename)
    
    private_key_parts = recover_private_key(signatures)
    
    # Afficher les parties récupérées
    for i, key_part in enumerate(private_key_parts):
        if key_part is not None:
            print(f"Key part {i}: {key_part.hex()}")
    
    # Si nous avons récupéré toutes les parties, construire la clé privée complète
    if all(k is not None for k in private_key_parts):
        complete_key = b''.join(private_key_parts)
        print("\nClé privée complète récupérée:")
        from_key = LamportPrivateKey(complete_key)
        print(from_key.dumps())

main()
````

````python
import binascii

def encrypter_texte(plaintext, mot_de_passe, cipher='aes-128-cbc'):
    pass_arg = 'pass:{}'.format(mot_de_passe)
    args = ['openssl', 'enc', '-' + cipher, '-base64', '-pass', pass_arg, '-pbkdf2']
    
    if isinstance(plaintext, str):
        plaintext = plaintext.encode('utf-8')

    result = subprocess.run(args, input=plaintext, stdout=subprocess.PIPE, stderr=subprocess.PIPE)

    message_erreur = result.stderr.decode()
    if message_erreur != '':
        raise OpensslError(message_erreur)

    return result.stdout.decode()

def chiffrer_avec_cle_publique(cle_publique, texte_en_clair):
    args = ['openssl', 'pkeyutl', '-encrypt', '-pubin', '-inkey', cle_publique]

    if isinstance(texte_en_clair, str):
        texte_en_clair = texte_en_clair.encode('utf-8')

    result = subprocess.run(args, input=texte_en_clair, stdout=subprocess.PIPE, stderr=subprocess.PIPE)

    message_erreur = result.stderr.decode()
    if message_erreur != '':
        raise OpensslError(message_erreur)

    return binascii.hexlify(result.stdout).decode()
````

TRAVAIL À FAIRE (l'emballage de l'activité pédagogique)
=======================================================
Vous obtiendrez un lot de transactions à traiter.  Plus précisément,
vous obtiendrez un dictionnaire avec :
  - identifier:   un identifiant de lot
  - transactions: une liste transactions

Vous devez déterminer si ces transactions sont valides.  Vous devrez
fournir une liste de booléens (un par transaction) indiquant si chacune
d'entre elle est valide.

Une fois que vous avez tenté votre chance avec un lot de transaction,
celui-ci est ``cramé'' et vous devez recommencer avec un autre.  Les
identifiants de lots sont pérennes et sont toujours valable après une
déconnexion, donc il est possible de travailler hors-ligne.

Pour vous entraîner, vous pouvez obtenir une transaction aléatoire,
son statut ainsi qu'une explication.  Un mélange de transaction valides
et invalides est renvoyé.  Récupérez plusieurs échantillons pour avoir
une idée des différents problèmes qui peuvent survenir.

FORMATTAGE
==========
Les données peuvent être affichées en JSON ou en YAML.  Le JSON est plus
facile à traiter automatiquement mais le YAML est plus human-friendly.
Le champ ``signature'' d'une transaction contient la signature du champ
``data'', qui contient les données de la transaction, sérialisées en JSON.

````python
import json
import binascii
from cryptography.hazmat.backends import default_backend
from cryptography.hazmat.primitives.asymmetric import padding, ec
from cryptography.hazmat.primitives import hashes
from cryptography.x509 import load_pem_x509_certificate
from cryptography.x509.oid import ExtensionOID
from cryptography.exceptions import InvalidSignature

def load_transactions_from_file(file_path):
    """
    Loads a batch of transactions from a JSON file.
    
    Args:
        file_path: Path to the JSON file containing transactions
        
    Returns:
        A list of transactions to validate
    """
    with open(file_path, 'r') as f:
        batch_data = json.load(f)
    return batch_data['batch']['transactions']

def verify_transaction(transaction):
    """
    Verifies if a transaction is valid according to EMV protocol requirements.
    
    Verification steps:
    1. Verify the transaction signature with the card's public key
    2. Verify the card certificate with the bank's public key
    3. Verify the bank certificate with the certification authority's public key
    4. Check if the bank certificate has CA authority
    5. Ensure transaction data contains card number and bank name
    
    Args:
        transaction: A dictionary containing transaction data
        
    Returns:
        1 if the transaction is valid, 0 if fraudulent
    """
    try:
        # Load certificates
        card_cert = load_pem_x509_certificate(
            transaction['card']['certificate'].encode(), 
            default_backend()
        )
        
        bank_cert = load_pem_x509_certificate(
            transaction['card']['bank']['certificate'].encode(), 
            default_backend()
        )
        
        # Load certification authority certificate
        with open('bdl/cetificat_UGLIX', "rb") as f:
            ca_cert = load_pem_x509_certificate(f.read(), default_backend())

        # Step 1: Verify transaction signature with card's public key
        card_cert.public_key().verify(
            binascii.unhexlify(transaction['signature']),
            transaction['data'].encode('utf-8'),
            ec.ECDSA(hashes.SHA256())
        )
        
        # Step 2: Verify card certificate with bank's public key
        bank_cert.public_key().verify(
            card_cert.signature,
            card_cert.tbs_certificate_bytes
        )
        
        # Step 3: Verify bank certificate with CA's public key
        ca_cert.public_key().verify(
            bank_cert.signature,
            bank_cert.tbs_certificate_bytes,
            padding.PKCS1v15(),
            bank_cert.signature_hash_algorithm
        )

        # Step 4: Check if bank certificate has CA authority
        if not bank_cert.extensions.get_extension_for_oid(ExtensionOID.BASIC_CONSTRAINTS).value.ca:
            raise Exception("Bank certificate does not have CA authority")
        
        # Step 5: Ensure transaction data contains card number and bank name
        if (transaction['card']['number'] not in transaction['data'] or 
            transaction['card']['bank']['name'] not in transaction['data']):
            raise Exception("Transaction data does not match card information")

        print(f"{transaction['card']['number']}: Valid transaction")
        return 1  # Valid transaction
        
    except Exception as e:
        print(f"{transaction['card']['number']}: Invalid transaction - {str(e) if str(e) else 'Verification failed'}")
        return 0  # Invalid transaction

def print_results(results):
    """
    Prints results in the expected format for submission.
    
    Args:
        results: List of verification results (1 for valid, 0 for invalid)
    """
    print("")
    for result in results:
        print(f"{result},", end='')

def main():
    """
    Main function that loads transactions and verifies each one.
    """
````

````python
from cryptography.hazmat.primitives import serialization
from cryptography.hazmat.backends import default_backend

def extraire_n_et_e(fichier_pem):
    """
    Extrait les composantes n (modulus) et e (exposant public) d'une clé publique RSA au format PEM.
    
    :param fichier_pem: Chemin vers le fichier PEM contenant la clé publique RSA.
    :return: Un tuple (n, e) où n est le modulus et e est l'exposant public.
    """
    try:
        # Charger la clé publique depuis le fichier PEM
        with open(fichier_pem, "rb") as fichier:
            cle_publique = serialization.load_pem_public_key(
                fichier.read(),
                backend=default_backend()
            )

        # Extraire n (modulus) et e (exposant public)
        n = cle_publique.public_numbers().n
        e = cle_publique.public_numbers().e

        return n, e

    except Exception as e:
        print(f"Erreur lors de la lecture du fichier PEM : {str(e)}")
        return None, None

````

````python
def creer_cle_privee_zkrause(fichier_pub_taylor, fichier_pub_zkrause, fichier_sortie):
    # Extraire n et e des deux clés publiques
    n_taylor, e_taylor = extraire_n_et_e(fichier_pub_taylor)
    n_zkrause, e_zkrause = extraire_n_et_e(fichier_pub_zkrause)
    
    # Vérifier que les deux utilisateurs partagent le même N
    if n_taylor != n_zkrause:
        print("Les utilisateurs ne partagent pas le même module N")
        return False
        
    n = n_taylor  # Le module partagé
    
    # Factoriser N en utilisant la clé privée de Taylor
    #p, q = factoriser_n(n, e_taylor, d_taylor)
    p, q = rsa.rsa_recover_prime_factors(n, e_taylor, d_taylor)
    
    if not p or p * q != n:
        print("Échec de factorisation du module N")
        return False
    
    # Calculer φ(N) = (p-1)(q-1)
    phi_n = (p - 1) * (q - 1)
    
    # Calculer d pour zkrause: d = e^(-1) mod φ(N)
    d_zkrause = pow(e_zkrause, -1, phi_n)
    
    # Créer les composantes pour la clé privée
    dmp1 = d_zkrause % (p - 1)
    dmq1 = d_zkrause % (q - 1)
    iqmp = pow(q, -1, p)
    
    # Créer l'objet clé privée
    private_numbers = rsa.RSAPrivateNumbers(
        p=p,
        q=q,
        d=d_zkrause,
        dmp1=dmp1,
        dmq1=dmq1,
        iqmp=iqmp,
        public_numbers=rsa.RSAPublicNumbers(e=e_zkrause, n=n)
    )
    
    private_key = private_numbers.private_key(default_backend())
    
    # Enregistrer en format PEM
    with open(fichier_sortie, "wb") as f:
        f.write(private_key.private_bytes(
            encoding=serialization.Encoding.PEM,
            format=serialization.PrivateFormat.PKCS8,
            encryption_algorithm=serialization.NoEncryption()
        ))
    
    print(f"Clé privée de zkrause créée avec succès dans {fichier_sortie}")
    print(f"Valeur d de zkrause (hex): {hex(d_zkrause)[2:]}")
    return True
````

````python
from cryptography.hazmat.primitives import serialization
from cryptography.hazmat.backends import default_backend
import binascii

def extraire_n_et_e(fichier_pem):
    """Extrait les composantes n et e d'une clé publique RSA"""
    try:
        with open(fichier_pem, "rb") as fichier:
            cle_publique = serialization.load_pem_public_key(
                fichier.read(),
                backend=default_backend()
            )
        n = cle_publique.public_numbers().n
        e = cle_publique.public_numbers().e
        return n, e
    except Exception as e:
        print(f"Erreur lors de la lecture du fichier PEM : {str(e)}")
        return None, None

def mod_inverse(a, m):
    """Calcule l'inverse modulaire de a modulo m"""
    # Utilise la relation de Bezout: a*u + m*v = gcd(a,m)
    # Si gcd(a,m) = 1, alors a*u ≡ 1 (mod m), donc u est l'inverse modulaire
    g, u, _ = xgcd(a, m)  # On utilise la fonction xgcd fournie par l'utilisateur
    if g != 1:
        raise Exception(f"L'inverse modulaire n'existe pas car gcd({a}, {m}) = {g}")
    return u % m

def dechiffrer_avec_deux_cles(ciphertext1_hex, ciphertext2_hex, key_file1, key_file2):
    """Déchiffre un message chiffré avec deux clés publiques partageant le même N"""
    # Extraire les clés publiques
    n1, e1 = extraire_n_et_e(key_file1)
    n2, e2 = extraire_n_et_e(key_file2)
    
    # Vérifier que les deux clés partagent le même N
    if n1 != n2:
        raise Exception("Les clés publiques ne partagent pas le même module N")
    
    n = n1  # Le module commun
    
    # Convertir les ciphertexts en entiers
    c1 = int(ciphertext1_hex, 16)
    c2 = int(ciphertext2_hex, 16)
    
    # Trouver les coefficients de Bezout: u*e1 + v*e2 = 1
    g, u, v = xgcd(e1, e2)
    if g != 1:
        raise Exception(f"Les exposants ne sont pas premiers entre eux: gcd({e1}, {e2}) = {g}")
    
    # Calcul du message original en utilisant la formule:
    # m = (c1^u * c2^v) mod n
    
    # Si u est négatif, calculer l'inverse modulaire
    if u < 0:
        c1_inv = mod_inverse(c1, n)
        part1 = pow(c1_inv, -u, n)
    else:
        part1 = pow(c1, u, n)
    
    # Si v est négatif, calculer l'inverse modulaire
    if v < 0:
        c2_inv = mod_inverse(c2, n)
        part2 = pow(c2_inv, -v, n)
    else:
        part2 = pow(c2, v, n)
    
    # Combiner les deux parties pour obtenir m
    m = (part1 * part2) % n
    
    # Convertir l'entier en bytes puis en texte
    try:
        # Convertir l'entier en bytes
        m_bytes = m.to_bytes((m.bit_length() + 7) // 8, byteorder='big')
        
        # Essayer de décoder en ASCII ou UTF-8
        try:
            m_text = m_bytes.decode('ascii')
        except UnicodeDecodeError:
            try:
                m_text = m_bytes.decode('utf-8')
            except UnicodeDecodeError:
                m_text = "Impossible de décoder en texte, voici l'hexadécimal: " + m_bytes.hex()
        
        return m_text
    except Exception as e:
        return f"Erreur lors du décodage: {str(e)}\nValeur numérique: {m}"

# Déchiffrement des deux messages
def dechiffrer_messages():
    # Message de leslieolson
    message1_hex = "0317be6babcda6938efaed74da699904b339f24afa68e501e58325c9ac285a0904f4c42d8e0a6253c8ca352916844645a4c7cd33c3158d84d6e7bf3df14207482c6f825ae7ce05ced78ae5f4b9fe853e18c53689fd849cd53c81e937b85db1ff2a59f41b52e900ac6f8e15b293eb1ca99e19541403c1987056eb26c36ab80211ac5857dbba80f5857318e56dd6dc9a948bdb37424cd2e5a04553fb61577f6cfeaf18b1cbecfbc1ca400a760b778b9aedd80229c9169e4df2a3df7bfa8be9d25a293ba6a2ed5de9d2802afcd36f2a9bb16deaa07f03c19e2779407457a316680b395a3bdcde0b69abf9607861e6121305ae9c0614f762b52f7adf2307b195d2f4"
    
    # Message de michaelwelch
    message2_hex = "8672f4cce023ee291b7dc017b06b85770e263716c6d2436c5a10d03c48799e11a1fa86a07b1c3a758366932a63dbf9098583d6fdeedfdb733cc92d9f356862931089c8f54d976fc751deeebbc834bf8f1e81d71baf7816d1b9c29a28301761dea74d22ea8b409d24b9b2c75e53e8d52c82fb9be03beced38a99b977d61089e49ac4b3081e0a18e372fc2caa0b66ca9f680f1acc1ac4e1500ee03f86ffda62e532dab7561fb4089b2369de97f558b210be488dae0d6f8c95caefbda8f7d58fe273c26eb9ac4b214559dd0babf9860061f713a5966b58a4521d2cd5dd78bdca3f12378a49e2163694e152715da4b5173c4826f84758e7885ddd7049a836cadd8de"
    
    # Utilisez les clés publiques des deux utilisateurs concernés
    texte_clair = dechiffrer_avec_deux_cles(
        message1_hex, 
        message2_hex, 
        "leslieolson-pk.pem",   # Remplacez par le chemin de la clé publique de leslieolson
        "michaelwelch-pk.pem"   # Remplacez par le chemin de la clé publique de michaelwelch
    )
    
    print("Message déchiffré:")
    print(texte_clair)
    
    # Essayer d'extraire un code digicode hexadécimal
    import re
    # Chercher une chaîne de caractères hexadécimaux (0-9, A-F)
    match = re.search(r'[0-9A-F]{4,}', texte_clair)
    if match:
        print("\nCode digicode trouvé:")
        print(match.group(0))

# Pour utiliser le script, assurez-vous d'avoir:
# 1. Les fichiers des clés publiques (leslieolson.pub et michaelwelch.pub)
# 2. La fonction xgcd déjà implémentée
# Puis appelez simplement:
dechiffrer_messages()
````

````python
from functools import reduce

# Paramètres
p = 2**64 - 59  # Premier de Mersenne
REQUIRED = {
    'R': 9,   # Degré 8 → 9 points
    'S': 17,  # Degré 16 → 17 points
    'T': 25   # Degré 24 → 25 points
}

def parse_mac(mac_hex):
    """Extrait les composants X, A, B, C d'un MAC hexadécimal"""
    mac_bytes = bytes.fromhex(mac_hex)
    X = int.from_bytes(mac_bytes[0:8], 'big') % p
    A = int.from_bytes(mac_bytes[8:16], 'big') % p
    B = int.from_bytes(mac_bytes[16:24], 'big') % p
    C = int.from_bytes(mac_bytes[24:32], 'big') % p
    return X, A, B, C

def lagrange_secret(points, prime):
    """Calcule le secret (terme constant) par interpolation de Lagrange"""
    secret = 0
    n = len(points)
    
    for i in range(n):
        xi, yi = points[i]
        numerator = yi
        denominator = 1
        
        for j in range(n):
            if j != i:
                numerator = (numerator * (-points[j][0])) % prime
                denominator = (denominator * (xi - points[j][0])) % prime
                
        inv_denominator = pow(denominator, prime-2, prime)
        term = (numerator * inv_denominator) % prime
        secret = (secret + term) % prime
    
    return secret

# Charger les MACs depuis le fichier
with open('flags.txt', 'r') as f:
    macs = [line.strip() for line in f if line.strip()]

# Préparer les données
data = {'R': [], 'S': [], 'T': []}
seen_x = set()

for mac in macs:
    try:
        X, A, B, C = parse_mac(mac)
        if X in seen_x:
            continue  # Éviter les X dupliqués
        seen_x.add(X)
        
        data['R'].append((X, A))
        data['S'].append((X, B))
        data['T'].append((X, C))
    except:
        continue

# Calculer les secrets
secrets = {}
for poly, required in REQUIRED.items():
    points = data[poly]
    if len(points) < required:
        print(f"Pas assez de points pour {poly} ({len(points)}/{required})")
        continue
    
    # Prendre les 'required' premiers points uniques
    selected = points[:required]
    secrets[poly] = lagrange_secret(selected, p)
    print(f"{poly}(0) = {secrets[poly]:016x}")

# Afficher les résultats
print("\nClés firmware reconstituées:")
print(f"R(0) (64 bits): {secrets.get('R', 'N/A'):016x}")
print(f"S(0) (64 bits): {secrets.get('S', 'N/A'):016x}")
print(f"T(0) (64 bits): {secrets.get('T', 'N/A'):016x}")
````

Pour s'en sortir, il faut absolument visualiser le principe général de 
fonctionnement de la fonction de hachage : la fonction de compression
traite les blocs de message les uns après les autres.  Plus précisément, 
le message à hacher s'écrit M = M_0 || M_1 || M_2 || ... et on pose :
    h[-1]  = IV
    h[i+1] = F(h[i], M_i)
La dernière valeur h[] produite est l'empreinte du message M.  Dans ces 
indications, on note :
    [h_1] ---X----> [h_2]
pour indiquer que, partant de l'état interne h_1, le fait de traiter la séquence
de blocs X aboutit à un nouvel état interne h_2.


[SUDO]>>> conseil zrr
Le programme de recherche de collision fonctionne de la façon suivante : 
- Il lit un "fichier préfixe" (éventuellement vide), dont la taille doit être
  un multiple de la taille des blocs.

- Le préfixe est haché, ce qui aboutit à une valeur de l'état interne h_prefix:
      [IV] ---Prefix---> [h_prefix] 

- À partir de là, un algorithme de recherche de collision est lancé. Cet
  algorithme produit deux séquences de deux blocs chacunes, (A_0, A_1) et
  (B_0, B_1), avec la propriété suivante :
      h_prefix ---A_0---> ??? ---A_1---> h_coll
      h_prefix ---B_0---> ??? ---B_1---> h_coll
Du coup, on a la collision :
    MD5(prefixe || A) == MD5(prefixe || B).
Notez que l'utilitaire n'inclut pas automatiquement le prefixe dans les
deux fichiers qu'il produit.


[SUDO]>>> conseil zrr
Pour mettre son nom d'utilisateur au début, il suffit de produire un préfixe
(de la bonne taille) qui commence par son nom d'utilisateur, et qui finisse
par n'importe quoi.


[SUDO]>>> conseil zrr
Pour mettre "h4ckm0d3" à la fin, ce n'est pas beaucoup plus compliqué.  L'astuce,
c'est qu'une fois qu'on a deux messages qui collisionnent, on peut les recycler
en rajoutant ce qu'on veut après.  En effet, on a :
    [IV] ---Prefix---> [h_prefix] ---A_0---> ??? ---A_1---> [h_coll]
    [IV] ---Prefix---> [h_prefix] ---B_0---> ??? ---B_1---> [h_coll]

Comme, de façon bien commode, les chaines : (Prefix || A) et (Prefix || B) ont
une taille qui est un multiple de la taille du bloc, tout ce qu'on pourrait 
rajouter après va aller dans un nouveau bloc, et ne va donc pas perturber la
collision :
    [IV] ---Prefix---> [h_prefix] ---A---> [h_coll] ---Suffix---> [h_suffix]
    [IV] ---Prefix---> [h_prefix] ---B---> [h_coll] ---Suffix---> [h_suffix]
Notez que le mécanisme de padding va entrer en action dans le suffixe, mais
que ça n'est pas un problème.


[SUDO]>>> conseil zrr
La partie un peu plus délicate consiste à produire non pas DEUX fichiers qui
collisionnent, mais QUATRE.  Pour cela, il faut utiliser la technique suivante,
inventée par Antoine Joux en 2004.  Partant d'une valeur de l'état interne
quelconque (disons h_0), on trouve une première collision :
    [h_0] ---A---> [h_1]
    [h_0] ---B---> [h_1]
Puis, partant de h_1, on trouve une deuxième collision :
    [h_1] ---C---> [h_2]
    [h_1] ---D---> [h_2]
On a donc :
    [h_0] ---A---> [h_1] ---C---> [h_2]
    [h_0] ---A---> [h_1] ---D---> [h_2]
    [h_0] ---B---> [h_1] ---C---> [h_2]
    [h_0] ---B---> [h_1] ---D---> [h_2]
Partant de l'état interne h_0, on a obtenu 4 séquences de 4 blocs qui
aboutissent au même état interne h_2 : AC, AD, BC, BD.

Concrètement, h_0 est l'état interne obtenu après avoir traité le préfixe.
Pour que l'utilitaire de recherche de collision cherche une collision à partir
de h_1, il faut lui donner un nouveau préfixe qui aboutisse à h_1. On peut lui
donner, par exemple, (Prefix || A).  Une fois qu'on a obtenu la 4-collision,
on peut rajouter "h4ckm0d3" à la fin des 4 messages qui collisionnent.

````python
import random
from sympy import factorint
import json

def multiplyList(myList):
    result = 1
    for x in myList:
        result *= x
    return result

def miller_rabin(n, k):
    if n in (2, 3):
        return True
    if n % 2 == 0:
        return False
    r, s = 0, n - 1
    while s % 2 == 0:
        r += 1
        s //= 2
    for _ in range(k):
        a = random.randrange(2, n - 1)
        x = pow(a, s, n)
        if x in (1, n - 1):
            continue
        for _ in range(r - 1):
            x = pow(x, 2, n)
            if x == n - 1:
                break
        else:
            return False
    return True

def bit_size(int):
    return (int.bit_length() + 7) // 8

def find_gen(p, list_div):
    while True:
        g = random.randrange(2, p-1)
        if all(pow(g, (p-1)//pi, p) != 1 for pi in list_div):
            return g

def rec_cert_pratt(list_div):
    res = []
    for div in list_div:
        cert_div = {'p': div}
        if div >= 1024:
            diviseur = []
            factors = factorint(div - 1)
            for d, exp in factors.items():
                diviseur.extend([d] * exp)
            cert_div['g'] = find_gen(div, diviseur)
            cert_div['pm1'] = rec_cert_pratt(diviseur)
        res.append(cert_div)
    return res

def main():
    a = int("b65faa15e48d84c86b7bada242a4192dfb8bbfd38f2eb207ea604b775d8475ab00224d3fa43a5f784305d8ba6690624f02ba14296a226e3ede8906eba14ce82123b169d77417b110d6b88aa7fbf60f3d342d58658ac66559cb01a4d78829d72cf58196859a19e7ce86b217ac9182175982d2f6989d087b08f4e4fc0c0c78367c", 16)
    b = a + 2**960

    liste_pk = [2]
    Q_prime = 2
    c = 20000
    len_b = bit_size(b)
    maxQprim = (b-a) // (c * len_b**2)

    while bit_size(maxQprim) != bit_size(Q_prime):
        pi = random.randrange(2, min(pow(2, 160), pow(2, 8 * (bit_size(maxQprim) - bit_size(Q_prime)))))
        if miller_rabin(pi, 10):
            Q_prime *= pi
            liste_pk.append(pi)

    p_k = random.randrange(a//Q_prime, b//Q_prime)
    while not miller_rabin(p_k, 1) or not miller_rabin(Q_prime*p_k+1, 1):
        p_k = random.randrange(a//Q_prime, b//Q_prime)

    liste_pk.append(p_k)
    p = multiplyList(liste_pk) + 1
    g = find_gen(p, liste_pk)

    cert_pratt = {'p': p, 'g': g, 'pm1': rec_cert_pratt(liste_pk)}
    print(cert_pratt)

    with open('flag24_tme.pratt.json', 'w') as f:
        json.dump(cert_pratt, f)

    print("Certificat généré et sauvegardé dans flag24_tme.pratt.json")
````

à executer dans sagemath

````python
from sage.all import *

def prime_as_sum_of_squares(p_hex):
    # Convert hex to integer
    p = Integer(int(p_hex, 16))
    
    # Check if p mod 4 = 1 (we'll assume primality given the context)
    if p % 4 != 1:
        return "Error: The number is not congruent to 1 modulo 4"
    
    print(f"Verified p ≡ 1 (mod 4): {p % 4 == 1}")
    
    # Find a value of x such that x² ≡ -1 (mod p)
    # For primes p ≡ 1 (mod 4), we can find a square root of -1
    x = Mod(-1, p).sqrt()
    x = int(x)
    
    # Apply Cornacchia's algorithm
    a, b = p, x
    while b*b > p:
        a, b = b, a % b
    
    c = isqrt(p - b*b)
    
    # Verify the result
    assert b*b + c*c == p
    
    return b, c

# The large prime in hex
p_hex = "e04e719bea1b3e97f1668af503e22d42dae24f77dd9064feafac99480477c5e0717830780751ced0be66eedf1953cc776d3dda0b1ade80dab4ddde291981c851965d8881f1a8977a0b146362201f89cc668559aff0af48d0f974ce159897d2a17ad89d459ed74fb14ec735fddc33c0d6e9066fde33ea1b54ae5ed1a163ebd779"

# Run the algorithm
a, b = prime_as_sum_of_squares(p_hex)
print(f"a = {a}")
print(f"b = {b}")
print(f"Verification: {a*a + b*b == int(p_hex, 16)}")
````

````python
# Définition de l'environnement et des variables
polynomial_q = x^64 + x^4 + x^3 + x + 1  # Polynôme irréductible pour GF(2^64)
finite_field = GF(2)['x']  
q_polynomial = finite_field(polynomial_q)   # Convertir q en polynôme dans F

# Conversion de la valeur hexadécimale en séquence de bits
hashpwd_hex = "5398adb0f347504b"
hashpwd_int = int.from_bytes(bytes.fromhex(hashpwd_hex), byteorder='big')
hashpwd_bin = bin(hashpwd_int)[2:].zfill(64)  # Binaire avec remplissage pour 64 bits

# Création du polynôme S(x) à partir de la séquence binaire
s_x = " + ".join([f"x^{i}" for i, bit in enumerate(reversed(hashpwd_bin)) if bit == '1'])
s_polynomial = finite_field(s_x)

# Calcul de l'inverse de (x^4 + x^3 + x + 1) modulo Q(x)
polynomial_to_inverse = finite_field(x^4 + x^3 + x + 1)
inverse_polynomial = polynomial_to_inverse.inverse_mod(q_polynomial)

# Calcul de P(x) comme l'inverse de Q multiplié par S
result_polynomial = inverse_polynomial * s_polynomial

# Conversion de P(x) en chaîne binaire et ensuite en hexadécimal
coef_list = [int(coef) for coef in result_polynomial.list()]
binary_str = ''.join(str(bit) for bit in coef_list[::-1])
int_value = int(binary_str, 2)
hex_value = hex(int_value)[2:].zfill(16)  # Assurez-vous que la chaîne hexadécimale a 16 caractères

# Afficher le résultat final
print("USER : MKT01")
print(f"Valeur Hexadécimale Résultante: {hex_value}")
````

````python
from hashlib import sha256
def key_expansion(seed : bytes) -> bytes:
    """
    Renvoie 256 bits pseudo-aléatoires à partir de seed
    """
    state = seed
    output = b''
    for i in range(8):
        state = sha256(state).digest()
        output += state[:4]
    return output

def get_seed(iv_cible):
    for i in range(2**32):
        seed = i.to_bytes(4, 'big')
        iv = bytes.fromhex(iv_cible)
        if key_expansion(seed)[16:] == iv:
            return seed
    return None

def decrypt_with_key_iv(plaintext, key, iv):
    if isinstance(plaintext, str):
        plaintext = plaintext.encode('utf-8')
    
    key_hex = key.hex()
    iv_hex = iv.hex()
    
    result = subprocess.run([
        'openssl', 'enc','-d', '-aes-128-cbc', '-K', key_hex, '-iv', iv_hex, '-base64'
    ], input=plaintext, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    
    error_message = result.stderr.decode()
    if error_message:
        raise OpensslError(error_message)
    
    return result.stdout.decode().strip()
````

````python
def create_known_bits_file(partial_hex, output_file):
    with open(output_file, 'w') as f:
        bit_position = 0
        for char in partial_hex.replace(':', ''):
            if char != '?':
                # Convert hex char to 4 bits
                hex_val = int(char, 16)
                for bit_offset in range(3, -1, -1):  # From MSB to LSB
                    bit_val = (hex_val >> bit_offset) & 1
                    f.write(f"{bit_position} {bit_val}\n")
                    bit_position += 1
            else:
                # Skip 4 bits for unknown hex character
                bit_position += 4

# Example usage
create_known_bits_file(partial_p_hex.replace('\n',''), "known_bits_p.txt")
create_known_bits_file(partial_q_hex.replace('\n',''), "known_bits_q.txt")
````

````python
import gmpy2

def branch_and_prune_factorization(N, p_known_bits, q_known_bits, p_length=1024, q_length=1024):
    """
    Recover RSA factors p and q using the branch and prune algorithm.
    
    Args:
        N: The RSA modulus
        p_known_bits: Dictionary mapping bit positions to known values for p
        q_known_bits: Dictionary mapping bit positions to known values for q
        p_length: Expected bit length of p
        q_length: Expected bit length of q
    
    Returns:
        A tuple (p, q) if factorization succeeds, None otherwise
    """
    # Start with empty candidates for p and q
    candidates = [(0, 0, 0)]  # (p_candidate, q_candidate, bit_position)
    
    # Process bit by bit, from least significant to most significant
    for bit_pos in range(max(p_length, q_length)):
        new_candidates = []
        
        for p_candidate, q_candidate, current_pos in candidates:
            if current_pos != bit_pos:
                continue
                
            # What we know about the current bits
            p_bit_known = bit_pos in p_known_bits
            q_bit_known = bit_pos in q_known_bits
            
            # Possible values for the current bits
            p_bit_values = [p_known_bits[bit_pos]] if p_bit_known else [0, 1]
            q_bit_values = [q_known_bits[bit_pos]] if q_bit_known else [0, 1]
            
            # Try all possible combinations
            for p_bit in p_bit_values:
                for q_bit in q_bit_values:
                    # Update candidates with the new bit
                    new_p = p_candidate | (p_bit << bit_pos)
                    new_q = q_candidate | (q_bit << bit_pos)
                    
                    # Check if the modular constraint is satisfied
                    if (new_p * new_q) % (1 << (bit_pos + 1)) == N % (1 << (bit_pos + 1)):
                        # This branch is still valid, keep it
                        new_candidates.append((new_p, new_q, bit_pos + 1))
        
        # Replace the old candidates with the new ones
        candidates = new_candidates
        
        # Periodic pruning to avoid excessive branching
        if len(candidates) > 10000:
            print(f"Warning: Too many candidates at bit {bit_pos}. Pruning...")
            candidates = candidates[:10000]
        
        if bit_pos % 10 == 0:
            print(f"Processing bit {bit_pos}, candidates: {len(candidates)}")
            
        # If we have no candidates left, the factorization has failed
        if not candidates:
            print(f"No valid candidates at bit {bit_pos}")
            return None
    
    # Check the final candidates against the full N
    for p, q, _ in candidates:
        if p * q == N:
            return p, q
    
    return None
````

````python
def parse_partial_key(partial_key_hex):
    """
    Convert a partial key in hex format with ? for unknown bits to a dictionary of known bits.
    
    Args:
        partial_key_hex: String like "e0:86:??:e3" with ? for unknown hex digits
        
    Returns:
        Dictionary mapping bit positions to their values (0 or 1)
    """
    known_bits = {}
    bit_position = 0
    
    # Clean the input
    partial_key_hex = partial_key_hex.replace(':', '').replace(' ', '')
    
    # Process each character
    for char in reversed(partial_key_hex):  # Reverse to start from least significant bits
        if char != '?':
            # Convert hex char to 4 bits
            hex_val = int(char, 16)
            for bit_offset in range(4):  # From LSB to MSB within the hex digit
                bit_val = (hex_val >> bit_offset) & 1
                known_bits[bit_position] = bit_val
                bit_position += 1
        else:
            # Skip 4 bits for unknown hex character
            bit_position += 4
    
    return known_bits

# Main execution
def recover_rsa_key():
    # Your RSA modulus from the dump
    modulus_hex = "a8c626e18c59462a2ae5b59103de25c675c1a97c0cc4a498ea38bf56956973d91f26832b7e3ae795347f0309f2bff1bdceb496e7e5cec7df410e9dafb247d2a86b7c9a06be10c171b090430e823086d17852128ddd19b61041db87a4ea1c2e3692a3553493c7de0c122b85c5ce8fe4914d59331e9f9a1d07fc294c4559a835d2e20d2dddc56652b9e2bd9ebf3d7ca3458eec657f99f3f37b4b5a2fd6db786f2561fc6f2e5b0e43f4b5292844a1e05220e8b6e71eef66304c50ebcd373e1c98aa97209c365e50bd1b9b06fdece03a6f3335a7e44f1b98f465220bec8e42a55461bc3b9464c540706df205dde7168366ea6f62297a4cadef6c3fde20d431"
    modulus = int(modulus_hex, 16)
    
    # Parse your partial p and q
    partial_p_hex = """e0:86:e8:e3:f0:03:bb:4e:36:a9:59:d?:79:f1:6?:?6
    ?b:e?:cd:??:1b:d1:da:?c:7?:35:87:b8:bc:?4:0f:?1
    dc:fd:f4:ff:1d:?d:b?:55:1e:47:e7:cf:07:a?:b6:10
    c0:67:ac:?6:?1:3?:49:d2:?f:aa:aa:d5:e3:24:1a:9b"""
    
    partial_q_hex = """?0:6e:8f:f0:c3:e9:a5:7a:6f:2?:07:8b:83:16:c7:8d
    29:a7:69:?3:?a:3d:?2:44:13:e2:c4:6?:e2:b8:0a:9c
    22:1c:c2:8f:?4:20:58:cc:4b:?b:df:7?:?8:6?:2f:13
    99:d3:8?:e0:?2:d?:a9:5d:b4:a?:8a:8a:2e:d0:ae:2a"""
    
    # Clean up the input
    partial_p_hex = partial_p_hex.replace('\n', '')
    partial_q_hex = partial_q_hex.replace('\n', '')
    
    # Parse to get known bit positions
    p_known_bits = parse_partial_key(partial_p_hex)
    q_known_bits = parse_partial_key(partial_q_hex)
    
    print(f"Found {len(p_known_bits)} known bits for p")
    print(f"Found {len(q_known_bits)} known bits for q")
    
    # Estimate prime factor sizes (assuming they're roughly the same size)
    prime_size = modulus.bit_length() // 2
    
    # Run the factorization
    p, q = branch_and_prune_factorization(modulus, p_known_bits, q_known_bits, 
                                         prime_size, prime_size)
    
    if p and q:
        print("Factorization successful!")
        print(f"p = {p}")
        print(f"q = {q}")
        print(f"Verification: p*q == N: {p*q == modulus}")
        
        # Create RSA key
        phi = (p-1)*(q-1)
        e = 65537  # Standard RSA public exponent
        d = int(gmpy2.invert(e, phi))
        
        print(f"Private exponent d = {d}")
        return p, q, d
    else:
        print("Factorization failed")
        return None
````

````python
import base64
import binascii
import json
import subprocess
from cryptography.hazmat.primitives.asymmetric import padding
from cryptography.hazmat.primitives import hashes, serialization

# Load the encrypted data from JSON
encrypted_data = {"session-key": "281e15064bd00f0622a3d212eb0be8c249797998d083bdb45662925b3167939464db8c42a1dea4331d816ed084cfdc0cae702fa60af9a0c8586572bc8cc29b86c2fc89908a91c46e0623851cf32437c1b39414a8349234030792e3a31a31696fd4a4f3db3ef844ec1c341b35b27f14ac2d8bffc5112b3d3f06cdba2e85d2b1b59208164ad42e660bcc918b38abf2beea95a87ca3cff7fab9bd099ac4bc2c8a5fb4f82bb7b3ae760d9751b34afc0e5e70a7cc7e86a030465d6de7e309644919b2daa14d0b9d434f1bf440641c98d82b17a7957840bd469222caffaf8f5903dd84ff62d0444957baae94b24577a74e1d2823f7d70607809a443298a0e4afac742c", "ciphertext": "U2FsdGVkX1+uf3t1cq7hObaj4rAxfqInL2zZbksljmv2gZnXqke5HjUN+db3QSaf\n3yUI7Os45Mm5M4L46cdSDDm9JztzfG/4/yybDM5GOMBwR2qYqGsc4jSsp88mLrjS\nknD9PZsJ/34+oGwP9o71xuHJVs6kwicFolQXZvGVskBch+bENrkFnzvpgIAayhw6\nmWaJ6zg2TlUCjaxgwFZZKRbH/uFJelW17P/xoSLwaCqVB1vP2tbe/0cOlnXJAJ6E\nLQR44RjQl/CT7h7WVPcYSGT9v1si9hIex3nftobwb93gRg1id44ArXlF9FrET/C9\nFuzEOrtpscHkkcLAB2lVPO4OKWoQznMvcpodXvcetG8WLLBUNhnzYRK8cXIEMhSi\nAUzGr4pxZjBAWvB3c5Rq4w==\n"}

#encrypted_data = json.loads(encrypted_data_json)

class OpensslError(Exception):
    pass

def decrypt_with_private_key(private_key_file, encrypted_hex):
    """Decrypt using private key"""
    # Convert hex to binary
    encrypted_data = binascii.unhexlify(encrypted_hex)
    
    # Write to temporary file
    with open('encrypted_data.bin', 'wb') as f:
        f.write(encrypted_data)
    
    # Decrypt with OpenSSL
    args = ['openssl', 'pkeyutl', '-decrypt', '-inkey', private_key_file, '-in', 'encrypted_data.bin']
    
    result = subprocess.run(args, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    
    message_error = result.stderr.decode()
    if message_error != '':
        raise OpensslError(message_error)
    
    return result.stdout.decode('utf-8')

def decrypt_text(encrypted_text, password, cipher='aes-128-cbc'):
    """Decrypt text using the password"""
    pass_arg = 'pass:{}'.format(password)
    
    # Store encrypted data in temporary file
    with open('encrypted_text.bin', 'w') as f:
        f.write(encrypted_text)
    
    args = ['openssl', 'enc', '-d', '-' + cipher, '-base64', '-pass', pass_arg, '-pbkdf2', '-in', 'encrypted_text.bin']
    
    result = subprocess.run(args, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    
    message_error = result.stderr.decode()
    if message_error != '':
        raise OpensslError(message_error)
    
    return result.stdout.decode('utf-8')

# Main decryption process
def decrypt_hybrid_message(private_key_path):
    # 1. Extract encrypted session key and ciphertext
    session_key_encrypted = encrypted_data["session-key"]
    ciphertext = encrypted_data["ciphertext"]
    
    # 2. Decrypt session key using private key
    try:
        session_key = decrypt_with_private_key(private_key_path, session_key_encrypted)
        print(f"Successfully decrypted session key: {session_key}")
    except OpensslError as e:
        print(f"Failed to decrypt session key: {e}")
        return None
    
    # 3. Decrypt ciphertext using session key
    try:
        plaintext = decrypt_text(ciphertext, session_key)
        print("\nDecrypted message:")
        print("------------------")
        print(plaintext)
        return plaintext
    except OpensslError as e:
        print(f"Failed to decrypt ciphertext: {e}")
        return None

# Replace with your private key path
private_key_path = "private_key.pem"
decrypt_hybrid_message(private_key_path)
````
